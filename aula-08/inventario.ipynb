```python
inventario = {}
for produto, quantidade in [("teclado", 10), ("rato", 25), ("monitor", 4)]:
    inventario[produto] = quantidade
atualizar = "rato"
if atualizar in inventario:
    inventario[atualizar] = 30
print("Inventário:")
for produto, qtd in inventario.items():
    print(f"{produto}: {qtd}")
```
