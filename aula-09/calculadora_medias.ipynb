```python
def media(notas):
    return sum(notas) / len(notas)

def classifica(media_final, minima=10):
    if media_final >= minima:
        return "Aprovado"
    return "Reprovado"

notas = []
for nota in [14, 9, 16]:
    notas.append(nota)
m = media(notas)
print(f"Média: {m:.1f}")
print(classifica(m))
```
