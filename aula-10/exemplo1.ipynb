```python
# Âmbito das variáveis: o que é criado dentro da função só existe dentro dela
def calcula_total(precos):
    subtotal = sum(precos)
    imposto = subtotal * 0.23
    return subtotal + imposto

t = calcula_total([10, 20, 5])
print(f"{t:.2f}")
# print(subtotal)  # NameError: name 'subtotal' is not defined (subtotal é local)
```
