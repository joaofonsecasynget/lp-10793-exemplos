```python
# Uma função como argumento de outra função
def dobro(x):
    return x * 2

def quadrado(x):
    return x ** 2

def aplica_a_todos(lista, funcao):
    resultado = []
    for item in lista:
        resultado.append(funcao(item))
    return resultado

numeros = [1, 2, 3, 4]
print(aplica_a_todos(numeros, dobro))
print(aplica_a_todos(numeros, quadrado))
```
