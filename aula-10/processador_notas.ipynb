```python
# Atividade: processador de notas (aqui com valores fixos; na aula, os dados vêm de input())
def classifica(nota):
    if nota >= 10:
        return "Aprovado"
    return "Reprovado"

notas = {"Ana": 14.0, "Bruno": 9.5, "Carla": 18.0}
resultados = {nome: classifica(nota) for nome, nota in notas.items()}
for nome, situacao in resultados.items():
    print(f"{nome}: {situacao}")
```
