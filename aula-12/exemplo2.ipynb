````python
contador = 0
with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        contador += 1
        print(f"{contador}. {linha.strip()}")
print(f"Total: {contador} produtos")
````

**Explicação:**

- O ciclo dá uma volta por linha: contador passa a 1, 2 e 3.
- linha.strip() tira o \n, por isso cada produto fica na sua linha, sem linhas em branco.
- Sem o .strip(), aparece uma linha em branco depois de cada produto.
- Extensão: print(texto) mostra as três linhas e uma linha em branco no fim (o último \n do ficheiro).
