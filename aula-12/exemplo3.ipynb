**Extensão (opcional):** troca "fruta\n" por um produto pedido com input("Produto: ") e acrescenta-o com f.write(produto + "\n")

Com `"a"`, cada execução acrescenta uma linha no fim. A célula seguinte corre duas vezes o acrescentar, para se ver `fruta` duas vezes:

````python
for _ in range(2):
    with open("compras.txt", "a", encoding="utf-8") as f:
        f.write("fruta\n")

with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        print(linha.strip())
````

Com `"w"`, o ficheiro é apagado e escrito de novo: só fica `fruta`.

````python
with open("compras.txt", "w", encoding="utf-8") as f:
    f.write("fruta\n")

with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        print(linha.strip())
````

Repor a lista (exercício 1) e **extensão** com `input()` (aqui com o produto «queijo» como exemplo):

````python
with open("compras.txt", "w", encoding="utf-8") as f:
    f.write("pão\n")
    f.write("leite\n")
    f.write("ovos\n")

produto = input("Produto: ")
with open("compras.txt", "a", encoding="utf-8") as f:
    f.write(produto + "\n")

with open("compras.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

**Explicação:**

- Com "a", cada execução acrescenta uma linha no fim: depois de duas execuções, fruta aparece duas vezes.
- Com "w", o ficheiro é apagado e escrito de novo: só fica fruta; pão, leite e ovos perdem-se.
- Extensão: produto = input("Produto: ") antes do with, e f.write(produto + "\n") no lugar de f.write("fruta\n").
