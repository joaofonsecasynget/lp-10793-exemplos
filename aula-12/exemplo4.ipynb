````python
try:
    with open("agenda.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("O ficheiro agenda.txt ainda não existe.")
print("Fim.")
````

**Explicação:**

- Sem try: FileNotFoundError: [Errno 2] No such file or directory: 'agenda.txt'.
- O ____ é FileNotFoundError. O erro é apanhado, aparece a mensagem e o programa continua até Fim.
