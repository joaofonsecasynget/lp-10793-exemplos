## Atividade: lista de tarefas guardada num ficheiro

1. Define mostrar_tarefas(): dentro de um try, abre tarefas.txt com "r" e mostra cada linha numerada, como no exercício 2.
2. No except FileNotFoundError: da mesma função, mostra "Ainda não há tarefas.".
3. Define acrescentar_tarefa(tarefa): abre tarefas.txt com "a" e escreve tarefa + "\n".
4. No fim: chama mostrar_tarefas(), pede nova = input("Nova tarefa: "), chama acrescentar_tarefa(nova) e volta a chamar mostrar_tarefas().
5. Executa duas vezes, com tarefas diferentes: na 2.ª execução, a primeira tarefa já lá está. Um escreve, o outro confere com os passos; depois troquem.

**Entrega:** aulas/aula-12.ipynb e tarefas.txt no repositório lp-10793, com commit e push

**Ponto de partida:** def mostrar_tarefas(): e, dentro, o try do exercício 4 com o ciclo do exercício 2

**Extensão (opcional):** em acrescentar_tarefa, se tarefa.strip() == "", mostra "Tarefa vazia: não foi guardada." e não escreve nada

```python
def mostrar_tarefas():
    try:
        with open("tarefas.txt", "r", encoding="utf-8") as f:
            numero = 0
            for linha in f:
                numero += 1
                print(f"{numero}. {linha.strip()}")
    except FileNotFoundError:
        print("Ainda não há tarefas.")

def acrescentar_tarefa(tarefa):
    with open("tarefas.txt", "a", encoding="utf-8") as f:
        f.write(tarefa + "\n")

mostrar_tarefas()
nova = input("Nova tarefa: ")
acrescentar_tarefa(nova)
mostrar_tarefas()
```

**Explicação:**

- mostrar_tarefas() junta o ciclo do exercício 2 e o try do exercício 4.
- acrescentar_tarefa() usa "a" para não apagar as tarefas anteriores.
- 1.ª execução (tarefa Estudar Python): ainda não há ficheiro, aparece a mensagem; depois, a lista tem 1 tarefa.
- 2.ª execução (tarefa Fazer commit e push): a lista já mostra a primeira tarefa e depois as duas.
- Extensão: if tarefa.strip() == "": print(…) e return, antes do with.
