**Extensão (opcional):** em vez do preço mais alto, imprime também o nome do produto mais caro

````python
import csv
with open("stock.csv", encoding="utf-8") as f:
    leitor = csv.DictReader(f)
    mais_caro = None
    for linha in leitor:
        preco = float(linha["preco"])
        if mais_caro is None or preco > mais_caro:
            mais_caro = preco
print(f"Preço mais alto: {mais_caro:.2f}")
````

**Extensão:**

````python
import csv
with open("stock.csv", encoding="utf-8") as f:
    leitor = csv.DictReader(f)
    mais_caro = None
    nome = None
    for linha in leitor:
        preco = float(linha["preco"])
        if mais_caro is None or preco > mais_caro:
            mais_caro = preco
            nome = linha["produto"]
print(f"Produto mais caro: {nome} ({mais_caro:.2f})")
````

**Explicação:**

- leitor = csv.DictReader(f) — lê por nome de coluna.
- preco = float(linha["preco"]) — converte o texto para número decimal.
- O resultado é 15.00 (a Mochila).
- Extensão: guardar também linha["produto"] numa variável sempre que se atualiza mais_caro.
