Confirma, abrindo stock_novo.csv, que ficam 6 linhas de produtos.

**Ponto de partida:** copia o código do diapositivo anterior e acrescenta uma segunda chamada a writerow()

````python
import csv
with open("stock.csv", encoding="utf-8") as f, \
     open("stock_novo.csv", "w", encoding="utf-8", newline="") as novo:
    leitor = csv.DictReader(f)
    escritor = csv.DictWriter(novo, fieldnames=["produto", "quantidade", "preco"])
    escritor.writeheader()
    for linha in leitor:
        escritor.writerow(linha)
    escritor.writerow({"produto": "Lapiseira", "quantidade": 25, "preco": "1.20"})
    escritor.writerow({"produto": "Régua", "quantidade": 40, "preco": "0.80"})

with open("stock_novo.csv", encoding="utf-8") as f:
    print(f.read())
````

**Explicação:**

- Basta repetir escritor.writerow({...}) com os dados de cada produto novo, uma chamada por produto.
- A ordem das chaves no dicionário não importa — o DictWriter usa sempre a ordem de fieldnames.
- O ficheiro final tem 1 linha de cabeçalho + 6 linhas de produtos = 7 linhas.
- Extensão: com input(), lembrar que os valores vêm como texto — não é preciso converter para escrever num CSV, só para fazer contas com eles.
