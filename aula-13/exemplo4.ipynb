## Exercício 4: escreve de raiz

```
# stock.xlsx — mesmos dados do stock.csv
# colunas: produto | quantidade | preco
# Caneta   | 50 | 0.60
# Caderno  | 30 | 2.50
# Borracha | 80 | 0.30
# Mochila  | 12 | 15.00
```

Escreve, de raiz, um pequeno programa que:
1. abre stock.xlsx com openpyxl;
2. percorre as linhas a partir da 2.ª;
3. conta quantos produtos têm menos de 20 unidades em stock;
4. imprime o resultado.

Testa no notebook: o resultado tem de ser 2 (Borracha tem 80 — não conta; Mochila tem 12 — conta).

**Entrega:** o teste no notebook, com o resultado 2

**Ponto de partida:** usa o exemplo do diapositivo anterior como ponto de partida para abrir e percorrer o ficheiro

**Extensão (opcional):** em vez de contar, imprime o nome de cada produto com menos de 20 unidades

````python
# Nota: o openpyxl não vem com o Python. Se o import falhar, instala-o no terminal do VS Code:
#   pip install openpyxl
# (sem permissões de administrador: pip install --user openpyxl)
import openpyxl
wb = openpyxl.load_workbook("stock.xlsx")
ws = wb.active
n = 0
for linha in ws.iter_rows(min_row=2, values_only=True):
    produto, quantidade, preco = linha
    if quantidade < 20:
        n += 1
print(f"Produtos com menos de 20 unidades: {n}")
````

**Extensão:**

````python
for linha in ws.iter_rows(min_row=2, values_only=True):
    produto, quantidade, preco = linha
    if quantidade < 20:
        print(produto)
````

**Explicação:**

- O código percorre as linhas, verifica a quantidade de cada produto e conta as que são menores do que 20.
- Resultado: 2 (Borracha tem 80, não conta; Mochila tem 12, conta).
