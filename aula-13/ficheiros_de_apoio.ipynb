Corre esta célula uma vez: cria `stock.csv` e `stock.xlsx` na pasta.

```python
# Nota: o openpyxl não vem com o Python. Se o import falhar, instala-o no terminal do VS Code:
#   pip install openpyxl
# (sem permissões de administrador: pip install --user openpyxl)
import openpyxl

with open("stock.csv", "w", encoding="utf-8") as f:
    f.write("produto,quantidade,preco\n")
    f.write("Caneta,50,0.60\n")
    f.write("Caderno,30,2.50\n")
    f.write("Borracha,80,0.30\n")
    f.write("Mochila,12,15.00\n")

wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Stock"
ws.append(["produto", "quantidade", "preco"])
for linha in [("Caneta", 50, 0.60), ("Caderno", 30, 2.50), ("Borracha", 80, 0.30), ("Mochila", 12, 15.00)]:
    ws.append(linha)
wb.save("stock.xlsx")
print("Criados stock.csv e stock.xlsx")
```
