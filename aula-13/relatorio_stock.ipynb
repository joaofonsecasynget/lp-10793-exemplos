## Atividade: relatório de stock

1. No teu notebook, lê o ficheiro stock.csv (já criado no exercício 1) com csv.DictReader.
2. Para cada produto, calcula o valor total: quantidade × preço (lembra-te de converter para número).
3. Cria uma folha de cálculo nova com openpyxl, com as colunas produto, quantidade, preco e valor total.
4. Escreve uma linha por produto, com o valor total calculado.
5. Guarda a folha como stock_relatorio.xlsx.
6. Abre o ficheiro gerado (no Excel, no LibreOffice ou lendo-o de novo com openpyxl) e confirma os 4 produtos e os valores.

**Entrega:** o stock_relatorio.xlsx com os 4 produtos e o valor total de cada um

**Ponto de partida:** usa o exemplo do diapositivo 9 (ler CSV) e do diapositivo 12 (escrever XLSX) como ponto de partida

**Extensão (opcional):** acrescenta, no fim da folha, uma última linha com o valor total de todo o stock

```python
# Nota: o openpyxl não vem com o Python. Se o import falhar, instala-o no terminal do VS Code:
#   pip install openpyxl
# (sem permissões de administrador: pip install --user openpyxl)
import csv
import openpyxl

relatorio = openpyxl.Workbook()
ws = relatorio.active
ws.title = "Relatório"
ws.append(["produto", "quantidade", "preco", "valor total"])
total_geral = 0

with open("stock.csv", encoding="utf-8") as f:
    leitor = csv.DictReader(f)
    for linha in leitor:
        quantidade = int(linha["quantidade"])
        preco = float(linha["preco"])
        valor = round(quantidade * preco, 2)
        total_geral += valor
        ws.append([linha["produto"], quantidade, preco, valor])

# Extensão: linha final com o total de todo o stock
ws.append(["TOTAL", None, None, round(total_geral, 2)])
relatorio.save("stock_relatorio.xlsx")

conferir = openpyxl.load_workbook("stock_relatorio.xlsx")
for linha in conferir.active.iter_rows(values_only=True):
    print(linha)
```

**Explicação:**

- Caneta: 50 × 0,60 = 30,00
- Caderno: 30 × 2,50 = 75,00
- Borracha: 80 × 0,30 = 24,00
- Mochila: 12 × 15,00 = 180,00
