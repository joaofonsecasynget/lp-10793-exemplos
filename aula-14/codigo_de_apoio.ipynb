# Código de apoio da aula 14

Corre esta célula uma vez (e de novo sempre que quiseres repor os dados originais): cria `escola.db` com a tabela `alunos`.

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("""
CREATE TABLE IF NOT EXISTS alunos (
    id INTEGER PRIMARY KEY,
    nome TEXT,
    nota REAL
)
""")
cursor.execute("DELETE FROM alunos")
alunos = [(1, "Ana", 16.0), (2, "Bruno", 12.5), (3, "Carla", 18.0), (4, "Duarte", 9.0)]
cursor.executemany("INSERT INTO alunos (id, nome, nota) VALUES (?, ?, ?)", alunos)
ligacao.commit()
ligacao.close()
print("Base de dados escola.db criada com 4 alunos")
```
