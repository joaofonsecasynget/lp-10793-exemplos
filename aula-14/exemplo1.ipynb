**Extensão:**

````python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota < 14")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
````

**Explicação:**

- O código imprime 2 linhas: ('Ana', 16.0) e ('Carla', 18.0) — só quem tem nota 14 ou mais.
- Cada linha vem como um tuplo, pela ordem das colunas pedidas no SELECT (nome, nota).
- Extensão: com < 14, vêm ('Bruno', 12.5) e ('Duarte', 9.0).
