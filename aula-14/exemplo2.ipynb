Substitui cada ____ para listar os alunos com nota abaixo de 10, do mais baixo para o mais alto.

A primeira e a terceira lacuna são palavras-chave de SQL; a segunda é ASC ou DESC; a última é o valor a usar no lugar do "?".

Testa no notebook: só deve aparecer Duarte.

**Entrega:** o teste no notebook, com só o Duarte no resultado

**Ponto de partida:** usa o exemplo do diapositivo anterior como modelo para as lacunas

**Extensão (opcional):** acrescenta uma segunda condição com AND, para excluir também quem tem nota 0

````python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota < ? ORDER BY nota ASC", (10,))
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
````

**Extensão:**

````python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota < ? AND nota > 0 ORDER BY nota ASC", (10,))
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
````

**Explicação:**

- A consulta fica: "SELECT nome, nota FROM alunos WHERE nota < ? ORDER BY nota ASC", (10,)
- Só o Duarte (9.0) tem nota abaixo de 10, por isso é o único resultado.
- Extensão: WHERE nota < ? AND nota > 0.
