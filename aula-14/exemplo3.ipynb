**Entrega:** o teste no notebook, com os dois alunos novos na listagem final

**Ponto de partida:** copia o exemplo e troca execute(INSERT...) por executemany(INSERT..., [tuplo1, tuplo2])

**Extensão (opcional):** antes de inserir, consulta o maior id já usado com SELECT MAX(id) FROM alunos, para não repetires um id

````python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
# Extensão: confirmar o maior id já usado
cursor.execute("SELECT MAX(id) FROM alunos")
maior = cursor.fetchone()[0]
print(f"Maior id: {maior}")

cursor.executemany("INSERT INTO alunos (id, nome, nota) VALUES (?, ?, ?)",
                   [(maior + 1, "Francisco", 13.0), (maior + 2, "Gabriela", 17.0)])
ligacao.commit()
cursor.execute("SELECT nome, nota FROM alunos")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
````

**Explicação:**

- cursor.executemany("INSERT INTO alunos (id, nome, nota) VALUES (?, ?, ?)", [(6, "Francisco", 13.0), (7, "Gabriela", 17.0)])
- Um só executemany() substitui dois execute() repetidos.
- Extensão: SELECT MAX(id) FROM alunos devolve um tuplo com um valor só, por exemplo (5,); cursor.fetchone()[0] dá esse número.
