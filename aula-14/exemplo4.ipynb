Escreve, de raiz, um pequeno programa que:
1. liga à base de dados, com row_factory = sqlite3.Row;
2. consulta todos os alunos com nota entre 10 e 14 (inclusive);
3. guarda o resultado numa lista de dicionários;
4. imprime o nome de cada um, um por linha.

Testa no notebook e confirma a lista com os teus próprios dados.

**Entrega:** o teste no notebook, com a lista de nomes impressa

**Ponto de partida:** usa o exemplo do diapositivo anterior como ponto de partida para a ligação e o row_factory

**Extensão (opcional):** em vez de só o nome, imprime também quantos alunos ficaram nesse intervalo, com len()

````python
import sqlite3
ligacao = sqlite3.connect("escola.db")
ligacao.row_factory = sqlite3.Row
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota >= 10 AND nota <= 14")
resultado = [dict(linha) for linha in cursor.fetchall()]
ligacao.close()
for aluno in resultado:
    print(aluno["nome"])
# Extensão
print(f"Alunos no intervalo: {len(resultado)}")
````

**Explicação:**

- A consulta fica: SELECT nome, nota FROM alunos WHERE nota >= 10 AND nota <= 14 — duas condições com AND, os dois limites incluídos.
- Com os dados do diapositivo 4, o resultado tem o Bruno (12.5); depende dos alunos que cada par acrescentou nos exercícios anteriores.
