## Atividade: relatório da turma

1. No teu notebook, liga à base de dados escola.db.
2. Consulta a média de todas as notas da turma, com a função SQL AVG(nota).
3. Consulta quantos alunos têm nota abaixo de 10, com COUNT(*) e WHERE.
4. Consulta o nome do aluno com a nota mais alta, com ORDER BY nota DESC e LIMIT 1.
5. Imprime as três respostas, cada uma com uma frase clara (por exemplo "Média da turma: 13.80").
6. Fecha a ligação no fim.

**Entrega:** o relatório impresso no notebook, com as três respostas

**Ponto de partida:** usa o exemplo do diapositivo 9 (execute e fetchall) como ponto de partida; AVG, COUNT e LIMIT são funções e cláusulas de SQL, usadas dentro da mesma string da consulta

**Extensão (opcional):** acrescenta uma quarta consulta: quantos alunos têm nota de 17 ou mais ("excelência")

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()

cursor.execute("SELECT AVG(nota) FROM alunos")
media = cursor.fetchone()[0]
print(f"Média da turma: {media:.2f}")

cursor.execute("SELECT COUNT(*) FROM alunos WHERE nota < 10")
print(f"Alunos com nota abaixo de 10: {cursor.fetchone()[0]}")

cursor.execute("SELECT nome FROM alunos ORDER BY nota DESC LIMIT 1")
print(f"Nota mais alta: {cursor.fetchone()[0]}")

# Extensão
cursor.execute("SELECT COUNT(*) FROM alunos WHERE nota >= 17")
print(f"Alunos com nota de 17 ou mais: {cursor.fetchone()[0]}")
ligacao.close()
```

**Explicação:**

- AVG(nota): "SELECT AVG(nota) FROM alunos" → cursor.fetchone()[0] dá a média.
- COUNT(*): "SELECT COUNT(*) FROM alunos WHERE nota < 10" → cursor.fetchone()[0] dá o número de alunos.
- LIMIT 1: "SELECT nome FROM alunos ORDER BY nota DESC LIMIT 1" → o primeiro (e único) resultado é o nome do aluno com a nota mais alta.
