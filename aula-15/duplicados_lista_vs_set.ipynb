## Atividade: qual é mais rápido?

1. No teu notebook, usa o código de apoio do professor (em aula-15.ipynb) para criar a lista numeros, com 3000 números ao acaso entre 1 e 5000.
2. Escreve a função tem_duplicados_lista(lista), que percorre a lista e usa uma lista vazia para guardar os valores já vistos, verificando cada novo valor com in.
3. Escreve a função tem_duplicados_set(lista), igual à primeira mas usando um set() em vez de uma lista para guardar os valores vistos.
4. Mede o tempo de cada uma com timeit.timeit(lambda: funcao(numeros), number=20).
5. Imprime os dois tempos e quantas vezes a mais rápida é mais rápida que a outra.
6. Nas tuas palavras, no notebook, explica porque é que uma é mais rápida: o que muda ao verificar "x in lista" e "x in set"?

**Entrega:** os dois tempos, a comparação e a explicação escrita no notebook

**Ponto de partida:** as duas funções têm a mesma estrutura: um ciclo for, um vistos vazio (lista ou set), um teste "in" e um .append()/.add()

**Extensão (opcional):** repete a comparação com uma lista de 30 000 números e vê se a diferença fica maior ou menor

```python
import timeit
import random

# Código de apoio: 3000 números ao acaso entre 1 e 5000
random.seed(15)
numeros = [random.randint(1, 5000) for _ in range(3000)]

def tem_duplicados_lista(lista):
    vistos = []
    for n in lista:
        if n in vistos:
            return True
        vistos.append(n)
    return False

def tem_duplicados_set(lista):
    vistos = set()
    for n in lista:
        if n in vistos:
            return True
        vistos.add(n)
    return False

t_lista = timeit.timeit(lambda: tem_duplicados_lista(numeros), number=20)
t_set = timeit.timeit(lambda: tem_duplicados_set(numeros), number=20)
print(f"Com lista: {t_lista:.4f} s")
print(f"Com set: {t_set:.4f} s")
print(f"A versão com set é cerca de {t_lista / t_set:.0f} vezes mais rápida")
```

**Porquê?** `n in vistos` numa lista obriga a olhar, um a um, para todos os elementos já vistos; num `set`, a verificação é direta, sem percorrer tudo. Os tempos exatos variam de computador para computador; o que interessa é a ordem de grandeza da diferença.

**Extensão:** com 30 000 números (sem duplicados garantidos, para obrigar a percorrer tudo).

```python
grandes = list(range(30000))
t_lista = timeit.timeit(lambda: tem_duplicados_lista(grandes), number=1)
t_set = timeit.timeit(lambda: tem_duplicados_set(grandes), number=1)
print(f"Com lista: {t_lista:.4f} s")
print(f"Com set: {t_set:.4f} s")
print(f"Diferença: cerca de {t_lista / t_set:.0f} vezes")
```

**Explicação:**

- tem_duplicados_lista: para cada número, verifica "n in vistos" numa lista — o Python tem de olhar, um a um, para todos os elementos já vistos.
- tem_duplicados_set: a mesma verificação, mas num set — um set sabe dizer se um valor lá está sem percorrer tudo, de forma muito mais direta.
- Por isso a versão com set ganha, e a diferença cresce quanto maior for a lista.
