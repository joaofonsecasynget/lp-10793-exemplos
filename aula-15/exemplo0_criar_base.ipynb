# Aula 15 — Exercício 0: cria a tua base de dados

**Exemplo (diapositivo de sintaxe):** cria `filmes.db` com a tabela `filmes`, de forma segura: `CREATE TABLE IF NOT EXISTS`, `DELETE` antes do `INSERT` (para poderes correr a célula outra vez), `commit()` e `try/finally` (a ligação fecha sempre, mesmo com erro, e a base não fica bloqueada).

```python
import sqlite3
ligacao = sqlite3.connect("filmes.db", timeout=5)
try:
    cursor = ligacao.cursor()
    cursor.execute("""CREATE TABLE IF NOT EXISTS filmes (
        id INTEGER PRIMARY KEY, titulo TEXT, ano INTEGER)""")
    cursor.execute("DELETE FROM filmes")
    cursor.executemany("INSERT INTO filmes (id, titulo, ano) VALUES (?, ?, ?)",
                       [(1, "Up", 2009), (2, "Coco", 2017)])
    ligacao.commit()
    cursor.execute("SELECT titulo, ano FROM filmes")
    print(cursor.fetchall())
finally:
    ligacao.close()
```

**Enunciado:** cria a base `musica.db` com a tabela `musicas` (`id`, `titulo`, `artista`), insere três músicas à tua escolha e mostra-as com um `SELECT`, uma por linha. Corre a célula duas vezes: o resultado tem de ser o mesmo, sem erros.

**Resolução:**

```python
import sqlite3
ligacao = sqlite3.connect("musica.db", timeout=5)
try:
    cursor = ligacao.cursor()
    cursor.execute("""CREATE TABLE IF NOT EXISTS musicas (
        id INTEGER PRIMARY KEY, titulo TEXT, artista TEXT)""")
    cursor.execute("DELETE FROM musicas")
    cursor.executemany("INSERT INTO musicas (id, titulo, artista) VALUES (?, ?, ?)",
        [(1, "Rosa", "Banda A"), (2, "Lume", "Banda B"), (3, "Mar", "Banda C")])
    ligacao.commit()
    cursor.execute("SELECT titulo, artista FROM musicas")
    for linha in cursor.fetchall():
        print(linha)
finally:
    ligacao.close()
```

**Extensão:** coluna `duracao` (INTEGER, em segundos) e ordenação da mais curta para a mais longa com `ORDER BY duracao`.

```python
import sqlite3
ligacao = sqlite3.connect("musica_ext.db", timeout=5)
try:
    cursor = ligacao.cursor()
    cursor.execute("""CREATE TABLE IF NOT EXISTS musicas (
        id INTEGER PRIMARY KEY, titulo TEXT, artista TEXT, duracao INTEGER)""")
    cursor.execute("DELETE FROM musicas")
    cursor.executemany("INSERT INTO musicas (id, titulo, artista, duracao) VALUES (?, ?, ?, ?)",
        [(1, "Rosa", "Banda A", 215), (2, "Lume", "Banda B", 187), (3, "Mar", "Banda C", 242)])
    ligacao.commit()
    cursor.execute("SELECT titulo, artista, duracao FROM musicas ORDER BY duracao")
    for linha in cursor.fetchall():
        print(linha)
finally:
    ligacao.close()
```

**Erros mais comuns**

- `OperationalError: database is locked` → outra ligação ficou aberta (uma célula que deu erro antes do `close()`): reinicia o kernel e usa sempre `try/finally`.
- `OperationalError: no such table` → a base que abriste não tem a tabela: foi criada noutra pasta ou o `CREATE TABLE` ainda não correu.
- `IntegrityError: UNIQUE constraint failed` → correste o `INSERT` duas vezes com o mesmo `id`: usa o `DELETE` antes.
