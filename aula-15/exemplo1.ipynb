**Ponto de partida:** usa o exemplo do diapositivo anterior como modelo, trocando a lista por uma string

**Extensão (opcional):** acrescenta um quarto print(next(letras)) e prevê o que acontece (dica: a string só tem 3 carateres)

````python
letras = iter("abc")
print(next(letras))
print(next(letras))
print(next(letras))
````

**Extensão:** um quarto `next()` levanta `StopIteration` (a string só tem 3 carateres).

````python
try:
    print(next(letras))
except StopIteration:
    print("StopIteration: já não há mais carateres")
````

**Explicação:**

- O código imprime a, b e c, um por linha — cada next() avança um carácter na string.
- Uma string funciona como qualquer sequência: iter() e next() não sabem nem querem saber se é uma lista ou um texto.
- Extensão: o quarto next(letras) levanta StopIteration, porque já não há mais carateres.
