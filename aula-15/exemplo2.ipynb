**Extensão (opcional):** escreve uma segunda função impares_ate(n), só com um while e um yield, sem copiar a primeira por inteiro

````python
def pares_ate(n):
    i = 0
    while i <= n:
        yield i
        i += 2

for numero in pares_ate(10):
    print(numero)
````

**Extensão:**

````python
def impares_ate(n):
    i = 1
    while i <= n:
        yield i
        i += 2

for numero in impares_ate(10):
    print(numero)
````

**Explicação:**

- i = 0 — começa no primeiro par.
- yield i — entrega o valor atual.
- i += 2 — avança de 2 em 2.
- Resultado com n = 10: 0, 2, 4, 6, 8, 10.
- Extensão: impares_ate(n) é igual, mas começa em i = 1.
