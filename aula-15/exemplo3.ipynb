## Exercício 3: altera um exemplo que funciona

```
quadrados = (n ** 2 for n in range(1, 6))
for q in quadrados:
    print(q)
```

Este código (do diapositivo anterior) gera os quadrados de 1 a 5.

Altera-o para gerar os cubos (n ** 3) dos números pares de 1 a 10.

Testa no notebook: tem de imprimir 8, 64, 216, 512 e 1000.

**Entrega:** o teste no notebook, com os 5 cubos impressos

**Ponto de partida:** copia o código do diapositivo anterior e muda só a expressão, o intervalo e a condição

**Extensão (opcional):** acrescenta, na mesma expressão geradora, uma condição extra para só incluir cubos acima de 100

````python
cubos_pares = (n ** 3 for n in range(1, 11) if n % 2 == 0)
for c in cubos_pares:
    print(c)
````

**Extensão:**

````python
cubos_grandes = (n ** 3 for n in range(1, 11) if n % 2 == 0 and n ** 3 > 100)
for c in cubos_grandes:
    print(c)
````

**Explicação:**

- cubos_pares = (n ** 3 for n in range(1, 11) if n % 2 == 0)
- A condição if n % 2 == 0 escolhe só os números pares, antes de elevar ao cubo.
- Resultado: 8, 64, 216, 512, 1000.
- Extensão: acrescentar and n ** 3 > 100 à condição — mas repara que n ** 3 tem de se calcular dentro da condição, não a seguir.
