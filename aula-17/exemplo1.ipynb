````python
class Livro:
    def __init__(self, titulo, paginas):
        self.titulo = titulo
        self.paginas = paginas

l1 = Livro("Python Fácil", 300)
l2 = Livro("Contos", 100)
l2.paginas = 120
print(l1.titulo, l1.paginas)
print(l2.titulo, l2.paginas)
print(l1.paginas + l2.paginas)
````

**Extensão:** `type()` mostra a classe do objeto.

````python
print(type(l1))
````

**Explicação:**

- l1 e l2 são dois objetos da mesma classe Livro: cada um tem os seus próprios atributos.
- l2.paginas = 120 muda só o l2; o l1 continua com 300.
- A soma dá 300 + 120 = 420.
- Extensão: type(l1) mostra <class '__main__.Livro'> — a classe Livro, definida no próprio notebook.
