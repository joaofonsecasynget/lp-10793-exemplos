## Exercício 3: altera um exemplo que funciona

```
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def com_iva(self):
        return self.preco * 1.23

cesto = [Produto("Pão", 1.20),
         Produto("Leite", 0.90)]
for p in cesto:
    print(f"{p.nome}: {p.com_iva():.2f} €")
```

Este código (do diapositivo de sintaxe) mostra o preço de cada produto com IVA.

Altera-o para cada produto ter também uma quantidade e o ciclo mostrar o total de cada produto: preço × quantidade (sem IVA).

Testa no notebook com 3 pães e 2 leites: tem de imprimir Pão: 3.60 € e Leite: 1.80 €.

**Entrega:** o teste no notebook, com os dois totais impressos

**Ponto de partida:** acrescenta quantidade ao __init__ (terceiro parâmetro) e troca o método com_iva() por total()

**Extensão (opcional):** depois do ciclo, soma os totais de todos os produtos e imprime o total do cesto

````python
class Produto:
    def __init__(self, nome, preco, quantidade):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade

    def total(self):
        return self.preco * self.quantidade

cesto = [Produto("Pão", 1.20, 3),
         Produto("Leite", 0.90, 2)]
for p in cesto:
    print(f"{p.nome}: {p.total():.2f} €")
````

**Extensão:**

````python
total_cesto = 0
for p in cesto:
    total_cesto += p.total()
print(f"Total do cesto: {total_cesto:.2f} €")
````

**Explicação:**

- O construtor passa a receber três parâmetros e guarda a quantidade em self.quantidade.
- O método total() devolve self.preco * self.quantidade.
- Cada objeto da lista tem os seus valores: 1.20 × 3 = 3.60 e 0.90 × 2 = 1.80.
- Extensão: um acumulador total_cesto, como nos ciclos com listas, dá 5.40 €.
