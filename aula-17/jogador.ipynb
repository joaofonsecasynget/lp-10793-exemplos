## Atividade: a classe Jogador

1. Cria a classe Jogador, com um construtor que recebe o nome e o guarda em self.nome.

2. No construtor, cria também o atributo pontos, que começa sempre em 0 (não é parâmetro).

3. Escreve o método ganhar(pontos), que soma os pontos recebidos ao atributo pontos.

4. Escreve o método mostrar(), que imprime, por exemplo, Ana tem 15 pontos.

5. Cria dois jogadores, Ana e Rui. A Ana ganha 10 e depois 5 pontos; o Rui ganha 7.

6. Chama mostrar() nos dois jogadores.

**Entrega:** a classe e o teste no notebook, com as duas linhas impressas

**Ponto de partida:** copia a classe Conta do diapositivo de sintaxe dos métodos e troca os nomes: titular → nome, saldo → pontos, depositar → ganhar

**Extensão (opcional):** imprime quem vence, com um if que compara os pontos dos dois jogadores

```python
class Jogador:
    def __init__(self, nome):
        self.nome = nome
        self.pontos = 0

    def ganhar(self, pontos):
        self.pontos += pontos

    def mostrar(self):
        print(f"{self.nome} tem {self.pontos} pontos")

j1 = Jogador("Ana")
j2 = Jogador("Rui")
j1.ganhar(10)
j1.ganhar(5)
j2.ganhar(7)
j1.mostrar()
j2.mostrar()
```

**Extensão:**

```python
if j1.pontos > j2.pontos:
    print(f"Vence: {j1.nome}")
else:
    print(f"Vence: {j2.nome}")
```

**Explicação:**

- self.pontos = 0 no construtor: todos os jogadores começam com 0 pontos, sem ser preciso passar o valor.
- ganhar() altera o atributo do próprio objeto com self.pontos += pontos.
- Cada objeto tem os seus pontos: a Ana fica com 15 e o Rui com 7.
- Extensão: os atributos usam-se fora da classe como qualquer valor, por exemplo num if.
