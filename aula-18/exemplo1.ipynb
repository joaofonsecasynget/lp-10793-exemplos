````python
class Termometro:
    def __init__(self):
        self._graus = 20

    def ajustar(self, graus):
        if graus >= -10 and graus <= 40:
            self._graus = graus
        else:
            print("Fora do intervalo")

    def graus(self):
        return self._graus

t = Termometro()
t.ajustar(25)
t.ajustar(90)
print(t.graus())
````

**Extensão:** o Python não impede o acesso ao `_graus` de fora; o sublinhado é só um aviso.

````python
t._graus = 90
print(t.graus())
````

**Explicação:**

- t.ajustar(25) aceita o valor, porque 25 está entre -10 e 40.
- t.ajustar(90) recusa o valor e imprime Fora do intervalo; _graus continua 25.
- print(t.graus()) mostra 25.
- Extensão: t._graus = 90 funciona e mostra 90 — o sublinhado é uma convenção, não uma proteção. Quem usa a classe não deve mexer em atributos com _.
