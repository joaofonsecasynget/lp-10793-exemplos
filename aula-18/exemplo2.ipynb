Substitui cada ____ para graus passar a ser uma propriedade: lê-se e altera-se sem parênteses, e o setter só aceita valores entre -10 e 40.

Testa no notebook: tem de imprimir 25 (o 90 é recusado).

**Ponto de partida:** usa a classe Conta dos diapositivos de @property e do setter como modelo

**Extensão (opcional):** acrescenta uma propriedade só de leitura, fahrenheit, que devolve graus * 9 / 5 + 32

````python
class Termometro:
    def __init__(self):
        self._graus = 20

    @property
    def graus(self):
        return self._graus

    @graus.setter
    def graus(self, valor):
        if valor >= -10 and valor <= 40:
            self._graus = valor

t = Termometro()
t.graus = 25
t.graus = 90
print(t.graus)
````

**Extensão:**

````python
class Termometro:
    def __init__(self):
        self._graus = 20

    @property
    def graus(self):
        return self._graus

    @graus.setter
    def graus(self, valor):
        if valor >= -10 and valor <= 40:
            self._graus = valor

    @property
    def fahrenheit(self):
        return self._graus * 9 / 5 + 32

t = Termometro()
t.graus = 25
print(t.fahrenheit)
````

**Explicação:**

- @property por cima do primeiro def graus: lê-se t.graus, sem parênteses.
- @graus.setter por cima do segundo def graus: corre quando se escreve t.graus = valor.
- self._graus = valor guarda o valor só quando passa na validação; o 90 é ignorado.
- Extensão: com 25 graus, fahrenheit dá 25 * 9 / 5 + 32 = 77.0.
