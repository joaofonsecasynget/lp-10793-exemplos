```
class Livro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def __str__(self):
        return f"{self.titulo}, de {self.autor}"

l = Livro("Mensagem", "Fernando Pessoa")
print(l)
```

Este código (do diapositivo de sintaxe do __str__) mostra o título e o autor de um livro.

Altera-o para o livro ter também o ano e o print() mostrar o ano entre parênteses, no fim.

Testa no notebook: tem de imprimir Mensagem, de Fernando Pessoa (1934).

**Entrega:** o teste no notebook, com a frase completa impressa

**Ponto de partida:** acrescenta ano ao __init__ e ({self.ano}) no fim da f-string do __str__

**Extensão (opcional):** acrescenta o __repr__ com o ano e testa com print([l])

````python
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def __str__(self):
        return f"{self.titulo}, de {self.autor} ({self.ano})"

l = Livro("Mensagem", "Fernando Pessoa", 1934)
print(l)
````

**Extensão:**

````python
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def __str__(self):
        return f"{self.titulo}, de {self.autor} ({self.ano})"

    def __repr__(self):
        return f"Livro('{self.titulo}', '{self.autor}', {self.ano})"

l = Livro("Mensagem", "Fernando Pessoa", 1934)
print([l])
````

**Explicação:**

- O construtor recebe o ano e guarda-o em self.ano.
- O __str__ devolve a f-string com o ano entre parênteses; o print(l) chama-o sozinho.
- Extensão: numa lista, o Python usa o __repr__ de cada objeto: [Livro('Mensagem', 'Fernando Pessoa', 1934)].
