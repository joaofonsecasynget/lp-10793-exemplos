## Atividade: a classe Produto

1. Cria a classe Produto, com um construtor que recebe nome e preco; guarda o nome em self.nome e o preço em self._preco.

2. Acrescenta a propriedade preco, com @property, que devolve self._preco.

3. Acrescenta o setter de preco, que só aceita valores maiores do que 0; senão, imprime Preço inválido.

4. Acrescenta o __str__, que devolve, por exemplo, Pão — 1.20 €.

5. Acrescenta o __eq__: dois produtos são iguais se tiverem o mesmo nome.

6. Testa: p1 = Produto("Pão", 1.20), p2 = Produto("Pão", 1.50), p1.preco = -3, print(p1) e print(p1 == p2).

**Entrega:** a classe e o teste no notebook, com as três linhas impressas

**Ponto de partida:** faz um passo de cada vez e corre a célula depois de cada um; copia o padrão da classe Conta

**Extensão (opcional):** acrescenta o __repr__, que devolve Produto('Pão', 1.2), e testa com print([p1, p2])

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self._preco = preco

    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, valor):
        if valor > 0:
            self._preco = valor
        else:
            print("Preço inválido")

    def __str__(self):
        return f"{self.nome} — {self._preco:.2f} €"

    def __eq__(self, outro):
        return self.nome == outro.nome

p1 = Produto("Pão", 1.20)
p2 = Produto("Pão", 1.50)
p1.preco = -3
print(p1)
print(p1 == p2)
```

**Extensão:**

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self._preco = preco

    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, valor):
        if valor > 0:
            self._preco = valor
        else:
            print("Preço inválido")

    def __str__(self):
        return f"{self.nome} — {self._preco:.2f} €"

    def __repr__(self):
        return f"Produto('{self.nome}', {self._preco})"

    def __eq__(self, outro):
        return self.nome == outro.nome

p1 = Produto("Pão", 1.20)
p2 = Produto("Pão", 1.50)
print([p1, p2])
```

**Explicação:**

- O preço fica em _preco e só se altera pelo setter, que recusa o -3: o p1 continua com 1.20.
- print(p1) usa o __str__: Pão — 1.20 €.
- p1 == p2 usa o __eq__, que só compara os nomes: True, mesmo com preços diferentes.
- Extensão: numa lista, o Python mostra o __repr__ de cada produto.
