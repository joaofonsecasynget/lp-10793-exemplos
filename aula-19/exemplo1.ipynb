````python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
````

**Extensão:** o `Veiculo` não tem `empinar()` (a mãe não herda da filha). Para o notebook não parar, a linha corre dentro de `try`/`except`.

````python
v = Veiculo("Fiat")
try:
    v.empinar()
except AttributeError as erro:
    print("AttributeError:", erro)
````

**Explicação:**

- A Mota herda do Veiculo o __init__ e o descrever(): m.descrever() mostra Yamaha com 4 rodas.
- Uma mota com 4 rodas está errado — vais corrigir isto no exercício 3, com super().
- m.empinar() existe na Mota: A Yamaha empinou!
- Extensão: AttributeError: 'Veiculo' object has no attribute 'empinar'.
