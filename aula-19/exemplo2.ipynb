**Extensão:**

````python
class Forma:
    def area(self):
        return 0

class Quadrado(Forma):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

class Circulo(Forma):
    def __init__(self, raio):
        self.raio = raio

    def area(self):
        return 3.14 * self.raio ** 2

class Triangulo(Forma):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura / 2

for f in [Quadrado(3), Circulo(2), Triangulo(4, 3)]:
    print(f.area())
````

**Explicação:**

- class Quadrado(Forma): a classe-mãe vai entre parênteses.
- def area(self): o mesmo nome do método da mãe — é isso que o sobrepõe.
- 3.14 * self.raio ** 2 usa o atributo do círculo: 3.14 * 4 = 12.56.
- O for chama area() em cada forma, e cada uma responde à sua maneira.
- Extensão: Triangulo(4, 3) dá 4 * 3 / 2 = 6.0.
