## Exercício 3: altera um exemplo que funciona

```
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
```

Este código (do exercício 1) diz que a mota tem 4 rodas.

Altera a classe Mota: dá-lhe um __init__ que chama o construtor do Veiculo com super() e depois muda self.rodas para 2.

Testa no notebook: tem de imprimir Yamaha com 2 rodas.

**Entrega:** o teste no notebook, com Yamaha com 2 rodas impresso

**Ponto de partida:** copia o padrão da classe Aluno do diapositivo de super(): primeiro super().__init__(...), depois o que muda

**Extensão (opcional):** sobrepõe também descrever() na Mota: chama super().descrever() e acrescenta print("(é uma mota)")

````python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def __init__(self, marca):
        super().__init__(marca)
        self.rodas = 2

    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
````

**Extensão:**

````python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def __init__(self, marca):
        super().__init__(marca)
        self.rodas = 2

    def descrever(self):
        super().descrever()
        print("(é uma mota)")

    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
````

**Explicação:**

- super().__init__(marca) corre o construtor do Veiculo: cria marca e rodas = 4.
- Logo a seguir, self.rodas = 2 corrige o valor só para as motas.
- O descrever() continua a ser o do Veiculo, mas agora lê rodas = 2.
- Extensão: super().descrever() reaproveita o método da mãe e a Mota acrescenta uma linha.
