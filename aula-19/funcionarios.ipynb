## Atividade: os pagamentos da empresa

1. Cria a classe Funcionario, com construtor (nome, salario) e o método pagamento(), que devolve o salário.

2. Cria a subclasse Gestor, com construtor (nome, salario, bonus): chama super().__init__(nome, salario) e guarda o bónus.

3. No Gestor, sobrepõe pagamento(): devolve super().pagamento() mais o bónus.

4. Cria a subclasse Estagiario, que só sobrepõe pagamento(): devolve metade do salário.

5. Cria a lista equipa com Funcionario("Rui", 1000), Gestor("Marta", 1500, 300) e Estagiario("Tiago", 800).

6. Num só for, imprime o nome e o pagamento de cada um e, no fim, o total.

**Entrega:** as três classes e o teste no notebook, com os quatro valores impressos

**Ponto de partida:** o Gestor segue o padrão da classe Aluno do diapositivo de super(); o for segue o diapositivo do polimorfismo

**Extensão (opcional):** acrescenta à classe Funcionario um __str__ que devolve o nome e o pagamento, e troca o print do for por print(f) — repara que serve para as três classes

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def pagamento(self):
        return self.salario

class Gestor(Funcionario):
    def __init__(self, nome, salario, bonus):
        super().__init__(nome, salario)
        self.bonus = bonus

    def pagamento(self):
        return super().pagamento() + self.bonus

class Estagiario(Funcionario):
    def pagamento(self):
        return self.salario / 2

equipa = [Funcionario("Rui", 1000), Gestor("Marta", 1500, 300),
          Estagiario("Tiago", 800)]
total = 0
for f in equipa:
    print(f"{f.nome}: {f.pagamento():.2f} €")
    total += f.pagamento()
print(f"Total: {total:.2f} €")
```

**Extensão:**

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def pagamento(self):
        return self.salario

    def __str__(self):
        return f"{self.nome}: {self.pagamento():.2f} €"

class Gestor(Funcionario):
    def __init__(self, nome, salario, bonus):
        super().__init__(nome, salario)
        self.bonus = bonus

    def pagamento(self):
        return super().pagamento() + self.bonus

class Estagiario(Funcionario):
    def pagamento(self):
        return self.salario / 2

equipa = [Funcionario("Rui", 1000), Gestor("Marta", 1500, 300),
          Estagiario("Tiago", 800)]
total = 0
for f in equipa:
    print(f)
    total += f.pagamento()
print(f"Total: {total:.2f} €")
```

**Explicação:**

- O Gestor reaproveita o construtor e o pagamento() da mãe com super(): 1500 + 300 = 1800.
- O Estagiario não tem __init__: herda o da mãe. Só muda o pagamento(): 800 / 2 = 400.
- O for chama f.pagamento() sem saber de que classe é cada objeto — isto é polimorfismo.
- Total: 1000 + 1800 + 400 = 3200.
- Extensão: o __str__ da mãe é herdado e chama o pagamento() de cada subclasse.
