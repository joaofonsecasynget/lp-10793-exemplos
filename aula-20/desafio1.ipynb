## Desafio 1: um decorador que limita as chamadas

1. Escreve o decorador limitar, para funções sem parâmetros, que deixa a função decorada correr no máximo 3 vezes.

2. Em cada chamada permitida, mostra Chamada N de 3 (N = 1, 2, 3) e só depois corre a função original.

3. A partir da 4.ª chamada, mostra Limite atingido e a função original já não corre.

4. Cada função decorada tem a sua própria contagem.

5. Aplica @limitar a enviar(), que imprime Mensagem enviada, e a guardar(), que imprime Dados guardados.

6. Testa: chama enviar() 4 vezes com um for e depois guardar() uma vez. Têm de aparecer 9 linhas, com um só Limite atingido, e guardar() começa em Chamada 1 de 3.

**Entrega:** o decorador e o teste no notebook, com as 9 linhas impressas.

**Extensão:** conta também as chamadas recusadas e mostra Limite atingido (recusadas: N); testa com 5 chamadas.

```python
def limitar(funcao):
    chamadas = []
    def nova():
        if len(chamadas) < 3:
            chamadas.append(1)
            print(f"Chamada {len(chamadas)} de 3")
            funcao()
        else:
            print("Limite atingido")
    return nova

@limitar
def enviar():
    print("Mensagem enviada")

@limitar
def guardar():
    print("Dados guardados")

for i in range(4):
    enviar()
guardar()
```

**Extensão:**

```python
def limitar(funcao):
    chamadas = []
    recusadas = []
    def nova():
        if len(chamadas) < 3:
            chamadas.append(1)
            print(f"Chamada {len(chamadas)} de 3")
            funcao()
        else:
            recusadas.append(1)
            print(f"Limite atingido (recusadas: {len(recusadas)})")
    return nova

@limitar
def enviar():
    print("Mensagem enviada")

for i in range(5):
    enviar()
```

**Explicação:**

- chamadas = [] é criada uma vez por função decorada: enviar e guardar têm cada uma a sua lista, por isso guardar() começa em Chamada 1 de 3.
- nova() lembra-se da lista do decorador (como multiplicar se lembrava do n) e junta-lhe um elemento em cada chamada permitida; len(chamadas) é a contagem.
- Com chamadas = 0 e chamadas += 1 dentro de nova, dá UnboundLocalError: cannot access local variable 'chamadas' where it is not associated with a value. O += torna chamadas uma variável local de nova, diferente da de limitar. Com a lista não há atribuição: append() altera a lista que já existe.
- Extensão: uma segunda lista, recusadas, que cresce no else; com 5 chamadas aparecem (recusadas: 1) e (recusadas: 2).
