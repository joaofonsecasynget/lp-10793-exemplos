## Desafio 2: produtos a partir de linhas de um CSV

Escreve as classes Produto e ProdutoSaldo para que o programa de teste (no fim da célula) funcione sem alterações:

- Produto(nome, preco, qtd), com o nome público e o preço e a quantidade protegidos (_);
- a propriedade total, com @property: preço × quantidade;
- @staticmethod valido(preco, qtd): True se os dois forem maiores do que 0;
- @classmethod de_linha(cls, linha): converte os textos da linha e devolve um objeto novo, ou None se não for válido;
- ProdutoSaldo herda de Produto e muda só o total (20 % de desconto); ProdutoSaldo.de_linha tem de criar um ProdutoSaldo sem escreveres outro de_linha.

**Entrega:** o programa a correr no notebook, a mostrar Caneta: 15.00 €, Mochila: 40.00 €, Ignorada: Régua, Estojo: 6.40 € e Total: 61.40 €.

**Extensão:** acrescenta um try/except ValueError em de_linha, para que a linha ["Caderno", "abc", "2", "normal"] seja ignorada em vez de dar erro.

```python
class Produto:
    def __init__(self, nome, preco, qtd):
        self.nome = nome
        self._preco = preco
        self._qtd = qtd

    @property
    def total(self):
        return self._preco * self._qtd

    @staticmethod
    def valido(preco, qtd):
        return preco > 0 and qtd > 0

    @classmethod
    def de_linha(cls, linha):
        preco = float(linha[1])
        qtd = int(linha[2])
        if not cls.valido(preco, qtd):
            return None
        return cls(linha[0], preco, qtd)


class ProdutoSaldo(Produto):
    @property
    def total(self):
        return self._preco * self._qtd * 0.8


linhas = [["Caneta", "1.50", "10", "normal"],
          ["Mochila", "25", "2", "saldo"],
          ["Régua", "-2", "3", "normal"],
          ["Estojo", "8", "1", "saldo"]]
soma = 0
for linha in linhas:
    if linha[3] == "saldo":
        p = ProdutoSaldo.de_linha(linha)
    else:
        p = Produto.de_linha(linha)
    if p is None:
        print("Ignorada:", linha[0])
    else:
        print(f"{p.nome}: {p.total:.2f} €")
        soma += p.total
print(f"Total: {soma:.2f} €")
```

**Extensão:**

```python
class Produto:
    def __init__(self, nome, preco, qtd):
        self.nome = nome
        self._preco = preco
        self._qtd = qtd

    @property
    def total(self):
        return self._preco * self._qtd

    @staticmethod
    def valido(preco, qtd):
        return preco > 0 and qtd > 0

    @classmethod
    def de_linha(cls, linha):
        try:
            preco = float(linha[1])
            qtd = int(linha[2])
        except ValueError:
            return None
        if not cls.valido(preco, qtd):
            return None
        return cls(linha[0], preco, qtd)


class ProdutoSaldo(Produto):
    @property
    def total(self):
        return self._preco * self._qtd * 0.8


print(Produto.de_linha(["Caderno", "abc", "2", "normal"]))
print(type(ProdutoSaldo.de_linha(["Estojo", "8", "1", "saldo"])))
```

**Explicação:**

- de_linha converte os textos com float() e int(), valida com cls.valido(…) — o mesmo que Produto.valido(…) — e devolve cls(…), ou None.
- Como devolve cls(…) e não Produto(…), ProdutoSaldo.de_linha cria um ProdutoSaldo: a filha herda o @classmethod sem o reescrever.
- ProdutoSaldo só sobrepõe a propriedade total: 25 × 2 × 0.8 = 40.00 e 8 × 1 × 0.8 = 6.40; a Caneta dá 1.50 × 10 = 15.00 e o total 61.40.
- A Régua tem preço -2: valido devolve False e o programa mostra Ignorada.
- Extensão: com o try/except, float("abc") deixa de parar o programa e de_linha devolve None; o type() confirma que ProdutoSaldo.de_linha devolve um ProdutoSaldo.
