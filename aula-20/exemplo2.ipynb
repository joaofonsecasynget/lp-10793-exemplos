**Ponto de partida:** usa o decorador anunciar do diapositivo anterior como modelo

**Extensão (opcional):** aplica o mesmo decorador a outra função, rodape(), que imprime Fim

````python
def moldura(funcao):
    def nova():
        print("=" * 10)
        funcao()
        print("=" * 10)
    return nova

@moldura
def titulo():
    print("RELATÓRIO")

titulo()
````

**Extensão:**

````python
def moldura(funcao):
    def nova():
        print("=" * 10)
        funcao()
        print("=" * 10)
    return nova

@moldura
def titulo():
    print("RELATÓRIO")

titulo()

@moldura
def rodape():
    print("Fim")

rodape()
````

**Explicação:**

- funcao() chama a função original (titulo) entre as duas linhas.
- return nova devolve a função nova, sem parênteses: não a chama, entrega-a.
- @moldura por cima de def titulo é o mesmo que titulo = moldura(titulo).
- Extensão: o mesmo decorador serve para qualquer função sem parâmetros, como rodape().
