## Exercício 3: altera um exemplo que funciona

```
class Temperatura:
    def __init__(self, celsius):
        self.celsius = celsius

    @staticmethod
    def c_para_f(c):
        return c * 9 / 5 + 32

    @classmethod
    def de_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

print(Temperatura.c_para_f(100))
t = Temperatura.de_fahrenheit(212)
print(t.celsius)
```

Este código (do diapositivo de sintaxe) cria temperaturas a partir de graus Fahrenheit.

Acrescenta um segundo @classmethod, de_kelvin(cls, k), que cria a temperatura a partir de kelvin: celsius = k - 273.15.

Testa no notebook com 300 kelvin e print(f"{t.celsius:.2f}"): tem de imprimir 26.85.

**Entrega:** o teste no notebook, com 26.85 impresso

**Ponto de partida:** copia o método de_fahrenheit e muda só o nome, o parâmetro e a conta

**Extensão (opcional):** acrescenta um @staticmethod c_para_k(c), que devolve c + 273.15, e testa com 25

````python
class Temperatura:
    def __init__(self, celsius):
        self.celsius = celsius

    @staticmethod
    def c_para_f(c):
        return c * 9 / 5 + 32

    @classmethod
    def de_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    @classmethod
    def de_kelvin(cls, k):
        return cls(k - 273.15)

t = Temperatura.de_kelvin(300)
print(f"{t.celsius:.2f}")
````

**Extensão:**

````python
class Temperatura:
    def __init__(self, celsius):
        self.celsius = celsius

    @staticmethod
    def c_para_f(c):
        return c * 9 / 5 + 32

    @staticmethod
    def c_para_k(c):
        return c + 273.15

    @classmethod
    def de_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    @classmethod
    def de_kelvin(cls, k):
        return cls(k - 273.15)

t = Temperatura.de_kelvin(300)
print(f"{t.celsius:.2f}")
print(Temperatura.c_para_k(25))
````

**Explicação:**

- @classmethod: o primeiro parâmetro é cls, a própria classe.
- return cls(k - 273.15) é o mesmo que Temperatura(k - 273.15): cria e devolve um objeto novo.
- 300 - 273.15 = 26.85; o :.2f mostra duas casas decimais.
- Extensão: c_para_k não precisa de self nem de cls; chama-se pela classe: 298.15.
