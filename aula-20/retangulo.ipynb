## Atividade: a classe Retangulo com decoradores

1. Cria a classe Retangulo, com construtor (largura, altura), que guarda os valores em self._largura e self._altura.

2. Acrescenta a propriedade area, com @property, que devolve largura × altura.

3. Acrescenta o @staticmethod valido(medida), que devolve True se a medida for maior do que 0.

4. Acrescenta o @classmethod quadrado(cls, lado), que devolve cls(lado, lado).

5. Testa: r = Retangulo(4, 3) e q = Retangulo.quadrado(5).

6. Imprime r.area, q.area e Retangulo.valido(-2).

**Entrega:** a classe e o teste no notebook, com os três valores impressos

**Ponto de partida:** segue a classe Temperatura do diapositivo de sintaxe; area escreve-se como o saldo da classe Conta

**Extensão (opcional):** copia o decorador moldura do exercício 2 e aplica-o a uma função mostrar_areas(), que imprime r.area e q.area

```python
class Retangulo:
    def __init__(self, largura, altura):
        self._largura = largura
        self._altura = altura

    @property
    def area(self):
        return self._largura * self._altura

    @staticmethod
    def valido(medida):
        return medida > 0

    @classmethod
    def quadrado(cls, lado):
        return cls(lado, lado)

r = Retangulo(4, 3)
q = Retangulo.quadrado(5)
print(r.area)
print(q.area)
print(Retangulo.valido(-2))
```

**Extensão:**

```python
def moldura(funcao):
    def nova():
        print("=" * 10)
        funcao()
        print("=" * 10)
    return nova

@moldura
def mostrar_areas():
    print(r.area, q.area)

mostrar_areas()
```

**Explicação:**

- @property: r.area lê-se sem parênteses e dá 4 × 3 = 12.
- @classmethod: Retangulo.quadrado(5) cria um retângulo 5 × 5; q.area dá 25.
- @staticmethod: valido(-2) não usa o objeto nem a classe; dá False.
- Extensão: o decorador moldura funciona com qualquer função sem parâmetros.
